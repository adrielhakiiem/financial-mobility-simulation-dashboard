## Step 1: Print original column names

```python
print("Income columns:", df_income.columns.tolist())
print("Poverty columns:", df_poverty.columns.tolist())
print("Gini columns:", df_gini.columns.tolist())
print("Amenities columns:", df_amenities.columns.tolist())
```

## Step 2: Standardize column names (strip whitespace and convert to lowercase)

```python
df_income.columns = df_income.columns.str.strip().str.lower()
df_poverty.columns = df_poverty.columns.str.strip().str.lower()
df_gini.columns = df_gini.columns.str.strip().str.lower()
df_amenities.columns = df_amenities.columns.str.strip().str.lower()
```

## Step 3: Print standardized column names to confirm

```python
print("Income columns (standardized):", df_income.columns.tolist())
print("Poverty columns (standardized):", df_poverty.columns.tolist())
print("Gini columns (standardized):", df_gini.columns.tolist())
print("Amenities columns (standardized):", df_amenities.columns.tolist())
```

## Step 4: Check overlapping years across all DataFrames

```python
# Get unique years from each DataFrame
years_income = set(df_income['year'].unique())
years_poverty = set(df_poverty['year'].unique())
years_gini = set(df_gini['year'].unique())
years_amenities = set(df_amenities['year'].unique())

print("Years in Income:", sorted(years_income))
print("Years in Poverty:", sorted(years_poverty))
print("Years in Gini:", sorted(years_gini))
print("Years in Amenities:", sorted(years_amenities))
```

```python
# Find overlapping years across all four DataFrames
overlapping_years = years_income & years_poverty & years_gini & years_amenities

print("\nOverlapping years across all four DataFrames:", sorted(overlapping_years))
```
